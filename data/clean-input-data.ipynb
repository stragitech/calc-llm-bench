# Clean input data
Exclude the six specified question IDs and save to `input_data_cleaned.csv`, leaving the source `input_data.csv` unchanged. Verify the output size and absence of excluded IDs.


```python
import csv
from pathlib import Path

# Run from this notebook's data directory or from the repository root.
csv_path = Path("input_data.csv")
if not csv_path.exists():
    csv_path = Path("data/input_data.csv")

output_path = csv_path.with_name("input_data_cleaned.csv")

excluded_ids = {
    "q28_sp23_problem09__04",
    "q9_sp22_problem05__02",
    "q18_fa22_problem04__11",
    "q5_sp22_problem05__04",
    "q21_fa22_problem01__11",
    "q26_sp23_problem05__04",
}

with csv_path.open(newline="", encoding="utf-8") as handle:
    reader = csv.DictReader(handle)
    fieldnames = reader.fieldnames
    rows = list(reader)

assert fieldnames and "ID" in fieldnames, "CSV must contain an ID column."
filtered_rows = [row for row in rows if row["ID"] not in excluded_ids]
removed_ids = {row["ID"] for row in rows if row["ID"] in excluded_ids}

with output_path.open("w", newline="", encoding="utf-8") as handle:
    writer = csv.DictWriter(handle, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(filtered_rows)

# Verify the saved file, including its size and the absence of excluded IDs.
with output_path.open(newline="", encoding="utf-8") as handle:
    saved_rows = list(csv.DictReader(handle))
remaining_ids = {row["ID"] for row in saved_rows} & excluded_ids
assert not remaining_ids, f"Excluded IDs still present: {sorted(remaining_ids)}"
assert saved_rows == filtered_rows, "Saved dataset differs from filtered rows."

print(f"Rows before: {len(rows)}")
print(f"Rows removed: {len(rows) - len(saved_rows)}")
print(f"Rows after: {len(saved_rows)}")
print(f"Excluded IDs removed: {len(removed_ids)} of {len(excluded_ids)}")
print(f"Excluded IDs remaining: {len(remaining_ids)}")
print(f"Cleaned dataset saved to: {output_path}")
```
